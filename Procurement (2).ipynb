## 1.3 O que é a Matriz de Kraljic?

A Matriz de Kraljic é uma ferramenta de gestão estratégica de compras que ajuda as empresas a classificar seus itens adquiridos com base em dois eixos principais:

Impacto no negócio (por exemplo: valor gasto, impacto no custo do produto vendido, criticidade para a operação);

Risco de fornecimento (por exemplo: dependência de poucos fornecedores, instabilidade de entrega, volatilidade de preços).

Essa classificação resulta em quatro quadrantes, que sugerem estratégias distintas para cada tipo de item:

| Quadrante         | Estratégia                          | Exemplo (aviação)                      |
|-------------------|--------------------------------------|----------------------------------------|
| **Itens Estratégicos**     | Parcerias de longo prazo               | Motores, sistemas hidráulicos críticos |
| **Itens Alavancáveis**     | Negociação agressiva de preços        | Pneus, óleos lubrificantes             |
| **Itens de Gargalo**       | Garantia de fornecimento (estoque)    | Componentes raros de difícil substituição |
| **Itens Não Críticos**     | Eficiência administrativa             | Materiais de escritório, EPIs          |

A seguir, apresento uma análise SWOT (Forças, Fraquezas, Oportunidades e Ameaças) para a iniciativa de otimização dos processos de procurement da MYK AIRLINES.

**Forças (Strengths)**





*Aplicação da Matriz de Kraljic*: A MYK AIRLINES está adotando uma abordagem estruturada e reconhecida internacionalmente para segmentar seus itens de compra e fornecedores, o que permite uma gestão mais estratégica e focada dos recursos e riscos.

**Fraquezas (Weaknesses)**


*Baixa previsibilidade na reposição de peças*: Isso gera dificuldades operacionais e pode comprometer a disponibilidade da frota, impactando diretamente a operação da companhia.

*Acúmulo de itens obsoletos no estoque*: Indica falhas na gestão de inventário e planejamento de compras, aumentando custos e imobilizando capital.

*Dependência de poucos fornecedores críticos*: Eleva o risco de desabastecimento e aumenta a vulnerabilidade da cadeia de suprimentos, especialmente em um setor tão sensível quanto o aéreo.

*Ausência de priorização estratégica no procurement até o momento*: Reflete processos ainda pouco maduros, que podem levar a decisões reativas e ineficientes.



**Oportunidades (Opportunities)**

*Diversificação e parcerias estratégicas*: A segmentação pela Matriz de Kraljic permite identificar oportunidades para diversificar fornecedores, negociar melhores condições e estabelecer parcerias que reduzam riscos e custos.

*Uso de tecnologias avançadas*: Automação e análise preditiva podem melhorar a gestão do estoque e a previsão de demanda, reduzindo atrasos e obsolescência.

*Mercado de fornecedores em desenvolvimento na América do Sul*: Pode haver fornecedores emergentes ou alternativas competitivas que ainda não foram exploradas, possibilitando renegociações e redução de custos.

*Adoção de estratégias específicas para cada cluster de itens*: Como leilões reversos para itens alavancáveis ou contratos de longo prazo para itens estratégicos, otimizando o custo e a segurança do fornecimento.




**Ameaças (Threats)**


*Volatilidade nos preços e riscos financeiros dos fornecedores*: Flutuações de preços e a possibilidade de falência ou incapacidade financeira de fornecedores críticos podem comprometer a cadeia de suprimentos e elevar o custo do produto vendido (CPV).

*Riscos logísticos e atrasos na cadeia de suprimentos*: Problemas externos, como instabilidade política, econômicas ou barreiras logísticas na América do Sul, podem impactar negativamente a entrega de componentes essenciais.

*Concorrência acirrada no setor aéreo*: Pressões para redução de custos e aumento da eficiência podem limitar a capacidade de investimento em melhorias no procurement.

*Regulamentações e exigências técnicas rigorosas*: O setor aeronáutico é altamente regulado, o que pode restringir a escolha de fornecedores e aumentar a complexidade da gestão de compras.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns



# Configurar estilo global
sns.set(style="whitegrid")

# Criar figura com 2 subplots (1 linha, 2 colunas)
fig, axes = plt.subplots(1, 2, figsize=(16, 6), gridspec_kw={'width_ratios': [2, 1]})

# -- Gráfico 1: Histograma --
sns.histplot(df['Lead_Time_Days'], bins=30, kde=True, color='skyblue', ax=axes[0])
axes[0].set_title('Distribuição do Tempo de Entrega (Dias)', fontsize=14)
axes[0].set_xlabel('Tempo de Entrega (Dias)', fontsize=12)
axes[0].set_ylabel('Frequência', fontsize=12)

# -- Gráfico 2: Box Plot com Anotações --
boxplot = sns.boxplot(
    y=df['Lead_Time_Days'],
    color='lightgreen',
    ax=axes[1]
)

# Calcular estatísticas
median = df['Lead_Time_Days'].median()
q1 = df['Lead_Time_Days'].quantile(0.25)
q3 = df['Lead_Time_Days'].quantile(0.75)
min_val = df['Lead_Time_Days'].min()
max_val = df['Lead_Time_Days'].max()

# Função para adicionar anotações ao box plot
def add_annotation(value, label, y_offset=0):
    boxplot.plot([0.1, 0.2], [value, value], color='black', linewidth=0.8, linestyle='--')  # linha até a anotação
    boxplot.annotate(label,
                     xy=(0.2, value),
                     xytext=(10, y_offset),
                     textcoords='offset points',
                     ha='left', va='center',
                     fontsize=10, color='black')

# Adicionar cada valor como anotação
add_annotation(min_val, f'Mínimo: {min_val:.1f}')
add_annotation(q1, f'Q1: {q1:.1f}', y_offset=-10)
add_annotation(median, f'Mediana: {median:.1f}', y_offset=0)
add_annotation(q3, f'Q3: {q3:.1f}', y_offset=10)
add_annotation(max_val, f'Máximo: {max_val:.1f}')

axes[1].set_title('Estatísticas do Tempo de Entrega', fontsize=14)
axes[1].set_ylabel('Tempo de Entrega (Dias)', fontsize=12)
axes[1].set_xticks([])  # remover rótulo do eixo X do boxplot

# Ajustar layout
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from scipy.stats import pointbiserialr, chi2_contingency
import warnings
warnings.filterwarnings('ignore')


# Definir um limiar para considerar risco de fornecimento alto
# Vamos considerar scores acima da mediana como alto risco
high_risk_threshold = df['Supply_Risk_Score'].median()
df['High_Supply_Risk'] = (df['Supply_Risk_Score'] > high_risk_threshold).astype(int)

# ============== VISUALIZAÇÕES ==============

# 1. Distribuição do Risco de Fornecimento
plt.figure(figsize=(10, 6))
sns.histplot(df['Supply_Risk_Score'], kde=True)
plt.axvline(high_risk_threshold, color='red', linestyle='--', label=f'Limiar de Alto Risco ({high_risk_threshold})')
plt.title('Distribuição do Risco de Fornecimento')
plt.xlabel('Score de Risco de Fornecimento')
plt.ylabel('Frequência')
plt.legend()
plt.tight_layout()
plt.show()

# 2. Análise de variáveis numéricas por nível de risco
numeric_features = ['Lead_Time_Days', 'Order_Volume_Units', 'Cost_per_Unit', 
                    'Profit_Impact_Score', 'Environmental_Impact']

for feature in numeric_features:
    plt.figure(figsize=(8, 5))
    sns.boxplot(x='High_Supply_Risk', y=feature, data=df)
    plt.title(f'{feature} por Nível de Risco')
    plt.xticks([0, 1], ['Baixo Risco', 'Alto Risco'])
    plt.tight_layout()
    plt.show()

# 3. Análise de variáveis categóricas por nível de risco
categorical_features = ['Supplier_Region', 'Single_Source_Risk', 'Kraljic_Category']

for feature in categorical_features:
    plt.figure(figsize=(10, 6))
    cat_counts = pd.crosstab(df[feature], df['High_Supply_Risk'], normalize='columns')
    cat_counts.plot(kind='bar')
    plt.title(f'Proporção de {feature} por Nível de Risco')
    plt.ylabel('Proporção')
    plt.xticks(rotation=45)
    plt.legend(['Baixo Risco', 'Alto Risco'])
    plt.tight_layout()
    plt.show()

# 4. Mapa de calor de correlação
numeric_cols = numeric_features + ['Supply_Risk_Score']
plt.figure(figsize=(10, 8))
correlation_matrix = df[numeric_cols].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Mapa de Calor de Correlação - Variáveis Numéricas')
plt.tight_layout()
plt.show()

# ============== ANÁLISES ESTATÍSTICAS ==============

# 1. Testes estatísticos para variáveis numéricas
numeric_stats = []

for feature in numeric_features:
    # Calcular médias para alto e baixo risco
    high_risk_mean = df[df['High_Supply_Risk'] == 1][feature].mean()
    low_risk_mean = df[df['High_Supply_Risk'] == 0][feature].mean()
    
    # Calcular diferença percentual
    diff_pct = ((high_risk_mean - low_risk_mean) / low_risk_mean) * 100
    
    # Correlação ponto-bisserial com o risco
    corr, p_value = pointbiserialr(df['High_Supply_Risk'], df[feature])
    
    # Armazenar resultados
    numeric_stats.append({
        'Feature': feature,
        'High_Risk_Mean': high_risk_mean,
        'Low_Risk_Mean': low_risk_mean,
        'Diff_Percent': diff_pct,
        'Correlation': corr,
        'P_Value': p_value,
        'Significant': p_value < 0.05
    })

numeric_stats_df = pd.DataFrame(numeric_stats)
numeric_stats_df = numeric_stats_df.sort_values('Correlation', ascending=False)

print("Estatísticas para variáveis numéricas:")
print(numeric_stats_df[['Feature', 'Diff_Percent', 'Correlation', 'Significant']])
print("\n")

# 2. Testes estatísticos para variáveis categóricas
categorical_stats = []

for feature in categorical_features:
    # Criar tabela de contingência
    contingency_table = pd.crosstab(df[feature], df['High_Supply_Risk'])
    
    # Realizar teste qui-quadrado
    chi2, p_value, dof, expected = chi2_contingency(contingency_table)
    
    # Calcular V de Cramer (medida de associação normalizada)
    n = contingency_table.sum().sum()
    phi2 = chi2 / n
    r, k = contingency_table.shape
    phi_coef = np.sqrt(phi2 / min(k-1, r-1))
    
    # Armazenar resultados
    categorical_stats.append({
        'Feature': feature,
        'Chi2': chi2,
        'P_Value': p_value,
        'Cramer_V': phi_coef,
        'Significant': p_value < 0.05
    })

categorical_stats_df = pd.DataFrame(categorical_stats)
categorical_stats_df = categorical_stats_df.sort_values('Cramer_V', ascending=False)

print("Estatísticas para variáveis categóricas:")
print(categorical_stats_df[['Feature', 'Cramer_V', 'P_Value', 'Significant']])
print("\n")

# 3. Importância das características usando Random Forest
# Preparar os dados para o modelo
X = pd.get_dummies(df.drop(['High_Supply_Risk', 'Supply_Risk_Score', 'Product_ID', 'Product_Name'], axis=1))
y = df['High_Supply_Risk']

# Dividir dados em treino e teste
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Treinar um modelo Random Forest
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# Avaliar o modelo
y_pred = rf_model.predict(X_test)
print("Desempenho do modelo Random Forest:")
print(classification_report(y_test, y_pred))

# Obter importância das características
feature_importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': rf_model.feature_importances_
})
feature_importance = feature_importance.sort_values('Importance', ascending=False)

# Visualizar importância das características
plt.figure(figsize=(10, 8))
sns.barplot(x='Importance', y='Feature', data=feature_importance.head(15))
plt.title('Importância das Características para Prever Alto Risco de Fornecimento')
plt.tight_layout()
plt.show()

# ============== ANÁLISE DE INSIGHTS CHAVE ==============

# 1. Identificar o tempo médio de entrega para alto e baixo risco
high_risk_lead_time = df[df['High_Supply_Risk'] == 1]['Lead_Time_Days'].mean()
low_risk_lead_time = df[df['High_Supply_Risk'] == 0]['Lead_Time_Days'].mean()
lead_time_diff = high_risk_lead_time - low_risk_lead_time

# 2. Comparar Single Source Risk por nível de risco
single_source_risk = pd.crosstab(df['Single_Source_Risk'], df['High_Supply_Risk'], normalize='columns')
single_source_high = single_source_risk[1]['Yes']
single_source_low = single_source_risk[0]['Yes']
single_source_diff = single_source_high - single_source_low

# 3. Relação entre região do fornecedor e risco
region_risk = pd.crosstab(df['Supplier_Region'], df['High_Supply_Risk'])
region_risk['Risk_Ratio'] = region_risk[1] / (region_risk[0] + region_risk[1])
highest_risk_region = region_risk.sort_values('Risk_Ratio', ascending=False).index[0]
highest_risk_ratio = region_risk.sort_values('Risk_Ratio', ascending=False)['Risk_Ratio'].iloc[0]

# 4. Relação entre categoria Kraljic e risco
kraljic_risk = pd.crosstab(df['Kraljic_Category'], df['High_Supply_Risk'])
kraljic_risk['Risk_Ratio'] = kraljic_risk[1] / (kraljic_risk[0] + kraljic_risk[1])
highest_risk_kraljic = kraljic_risk.sort_values('Risk_Ratio', ascending=False).index[0]
highest_risk_kraljic_ratio = kraljic_risk.sort_values('Risk_Ratio', ascending=False)['Risk_Ratio'].iloc[0]

# 5. Identificar impacto ambiental médio para alto e baixo risco
high_risk_env = df[df['High_Supply_Risk'] == 1]['Environmental_Impact'].mean()
low_risk_env = df[df['High_Supply_Risk'] == 0]['Environmental_Impact'].mean()
env_impact_diff = high_risk_env - low_risk_env

# ============== RELATÓRIO FINAL ==============

print("\n📌 CONCLUSÕES SOBRE OS FATORES ASSOCIADOS AO ALTO RISCO DE FORNECIMENTO")
print("=====================================================================================\n")
print("Esta análise utilizou abordagens estatísticas e machine learning para identificar")
print("os fatores mais associados ao alto risco na cadeia de fornecimento.\n")

print("Principais conclusões:")

# Feature importance do modelo Random Forest
top_features = feature_importance.head(3)['Feature'].tolist()
print(f"1. As variáveis com maior poder preditivo sobre o risco de fornecimento são: {', '.join(top_features)}")

# Correlação e diferença para Lead Time
lead_time_corr = numeric_stats_df[numeric_stats_df['Feature'] == 'Lead_Time_Days']['Correlation'].values[0]
print(f"2. Correlação entre tempo de entrega e risco de fornecimento: {lead_time_corr:.3f}")
print(f"   Tempo médio de entrega para produtos de baixo risco: {low_risk_lead_time:.1f} dias")
print(f"   Tempo médio de entrega para produtos de alto risco: {high_risk_lead_time:.1f} dias")
print(f"   Diferença: {lead_time_diff:.1f} dias ({(lead_time_diff/low_risk_lead_time)*100:.1f}% maior)")

# Single Source Risk
print(f"3. Proporção de produtos de fonte única em alto risco: {single_source_high*100:.1f}%")
print(f"   Proporção de produtos de fonte única em baixo risco: {single_source_low*100:.1f}%")
print(f"   Diferença: {(single_source_diff)*100:.1f} pontos percentuais")

# Região do fornecedor
print(f"4. Região com maior taxa de risco: {highest_risk_region} ({highest_risk_ratio*100:.1f}%)")

# Categoria Kraljic
print(f"5. Categoria Kraljic com maior taxa de risco: {highest_risk_kraljic} ({highest_risk_kraljic_ratio*100:.1f}%)")

# Impacto ambiental
print(f"6. Impacto ambiental médio em produtos de alto risco: {high_risk_env:.1f}")
print(f"   Impacto ambiental médio em produtos de baixo risco: {low_risk_env:.1f}")
print(f"   Diferença: {env_impact_diff:.1f} ({(env_impact_diff/low_risk_env)*100:.1f}% maior)\n")

print("Recomendações:")
print("- Implementar monitoramento reforçado para produtos com tempo de entrega longo")
print(f"- Desenvolver fornecedores alternativos para itens de fonte única, especialmente na região {highest_risk_region}")
print(f"- Priorizar estratégias de mitigação de risco para produtos na categoria {highest_risk_kraljic}")
print("- Considerar fatores ambientais na avaliação de risco, já que estão correlacionados com risco de fornecimento")
print("- Estabelecer limites de alerta para tempo de entrega e outros indicadores chave identificados")
print("- Revisar acordos com fornecedores nas regiões de maior risco e desenvolver planos de contingência")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# Ignorar warnings desnecessários
import warnings
warnings.filterwarnings('ignore')

# ============== CARREGAMENTO DE DADOS ==============

# Exibir informações básicas
print("📊 Informações iniciais do dataset:")
print(df[['Supply_Risk_Score', 'Profit_Impact_Score']].describe())

# ============== PRÉ-PROCESSAMENTO DOS DADOS ==============
# Selecionar variáveis relevantes para segmentação
X = df[['Supply_Risk_Score', 'Profit_Impact_Score']]

# Padronizar os dados (importante para K-Means)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# ============== AGRUPAMENTO COM K-MEANS ==============
# Aplicar K-Means com 4 clusters
kmeans = KMeans(n_clusters=4, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)

# Recuperar os centros dos clusters em escala original
centers = scaler.inverse_transform(kmeans.cluster_centers_)
centers_df = pd.DataFrame(centers, columns=['Supply_Risk_Score', 'Profit_Impact_Score'])

# Função para mapear clusters aos quadrantes da Matriz de Kraljic
def map_cluster_to_category(cluster_index, centers):
    supply_risk = centers.loc[cluster_index, 'Supply_Risk_Score']
    profit_impact = centers.loc[cluster_index, 'Profit_Impact_Score']

    # Dividir por medianas para definir limites entre "alto" e "baixo"
    risk_med = centers['Supply_Risk_Score'].median()
    profit_med = centers['Profit_Impact_Score'].median()

    if supply_risk > risk_med and profit_impact > profit_med:
        return 'Estratégico'
    elif supply_risk > risk_med and profit_impact <= profit_med:
        return 'Gargalo'
    elif supply_risk <= risk_med and profit_impact > profit_med:
        return 'Alavancável'
    else:
        return 'Não-crítico'

# Aplicar rotulagem
df['Kraljic_Category_Predicted'] = df['Cluster'].apply(lambda x: map_cluster_to_category(x, centers_df))

# ============== VISUALIZAÇÕES ==============
# Gráfico dos clusters originais
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Supply_Risk_Score', y='Profit_Impact_Score', hue='Cluster', data=df, palette='Set1', alpha=0.7)
plt.title('Clusters Originais - Risco vs Impacto')
plt.xlabel('Risco de Fornecimento')
plt.ylabel('Impacto no Lucro')
plt.tight_layout()
plt.show()

# Gráfico com categorias nomeadas (Matriz de Kraljic)
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Supply_Risk_Score', y='Profit_Impact_Score', hue='Kraljic_Category_Predicted', data=df, palette='Dark2', alpha=0.7)
plt.title('Segmentação por Matriz de Kraljic')
plt.xlabel('Risco de Fornecimento')
plt.ylabel('Impacto no Lucro')
plt.tight_layout()
plt.legend(title='Categoria')
plt.show()

# Mostrar centros dos clusters
print("\n🎯 Centros dos clusters (valores nas escalas originais):")
print(centers_df)

# Contagem de itens por categoria
print("\n📦 Distribuição de itens por categoria:")
print(df['Kraljic_Category_Predicted'].value_counts())

# Opcional: Usar PCA para visualização em 2D com mais contexto
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
df['PCA1'] = X_pca[:, 0]
df['PCA2'] = X_pca[:, 1]

plt.figure(figsize=(8, 6))
sns.scatterplot(x='PCA1', y='PCA2', hue='Kraljic_Category_Predicted', data=df, palette='Dark2', alpha=0.7)
plt.title('Visualização das Categorias via PCA')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.tight_layout()
plt.legend(title='Categoria')
plt.show()

# ============== RELATÓRIO FINAL ==============
print("\n📌 CONCLUSÕES SOBRE A SEGMENTAÇÃO POR MATRIZ DE KRALJIC")
print("=====================================================================================")
print("Esta análise utilizou técnicas de agrupamento (K-Means) para segmentar os itens de compra")
print("segundo a Matriz de Kraljic, com base nos critérios de risco de fornecimento e impacto no lucro.\n")
print("Quadrantes identificados:")
for cat in df['Kraljic_Category_Predicted'].unique():
    count = df[df['Kraljic_Category_Predicted'] == cat].shape[0]
    print(f"- {cat}: {count} produtos")

# Exemplo de insights por categoria
avg_cost = df.groupby('Kraljic_Category_Predicted')['Cost_per_Unit'].mean().round(2).to_dict()
avg_lead_time = df.groupby('Kraljic_Category_Predicted')['Lead_Time_Days'].mean().round(1).to_dict()

print("\n💰 Custo médio por categoria:")
for cat, cost in avg_cost.items():
    print(f"{cat}: R$ {cost}")

print("\n⏳ Tempo médio de entrega por categoria:")
for cat, lt in avg_lead_time.items():
    print(f"{cat}: {lt} dias")

print("\n💡 Recomendações:")
print("- Produtos estratégicos devem ter parcerias fortes com fornecedores e planos de contingência.")
print("- Produtos em gargalo exigem atenção redobrada, mesmo que não gerem muito lucro.")
print("- Produtos alavancáveis são boas oportunidades para economia através de negociação.")
print("- Produtos não-críticos podem ser automatizados ou comprados com menos frequência.")
print("- Revisar regularmente a classificação com novos dados para manter a estratégia atualizada.")
```

**E)Quais características do processo de compras estão mais associadas a compras ineficientes (alto custo e baixo ROI), e como podemos priorizar melhorias?**



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import f_oneway, chi2_contingency
import warnings
warnings.filterwarnings('ignore')



print("📊 Informações iniciais do dataset:")
print(df[['Kraljic_Category', 'Supply_Risk_Score', 'Profit_Impact_Score']].describe())

# ============== 1. COMPARAÇÃO DE MÉDIAS POR CATEGORIA ==============
numeric_features = ['Lead_Time_Days', 'Cost_per_Unit', 'Order_Volume_Units',
                    'Supply_Risk_Score', 'Profit_Impact_Score', 'Environmental_Impact']

print("\n🔹 Médias das variáveis por categoria Kraljic:")

category_means = df.groupby('Kraljic_Category')[numeric_features].mean().round(2)
print(category_means)

# Visualização
plt.figure(figsize=(12, 6))
category_means.plot(kind='bar', rot=0, title='Médias por Categoria Kraljic')
plt.ylabel('Valor Médio')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# ANOVA para cada variável numérica
anova_results = []
for feature in numeric_features:
    groups = [group[feature].values for name, group in df.groupby('Kraljic_Category')]
    f_stat, p_val = f_oneway(*groups)
    anova_results.append({
        'Feature': feature,
        'F_Statistic': f_stat,
        'P_Value': p_val,
        'Significant': p_val < 0.05
    })

anova_df = pd.DataFrame(anova_results)
print("\n✅ Resultados do teste ANOVA (diferenças entre médias):")
print(anova_df)

# ============== 2. TESTES DE ASSOCIAÇÃO PARA VARIÁVEIS CATEGÓRICAS ==============
categorical_features = ['Supplier_Region', 'Single_Source_Risk']

chi2_results = []
for feature in categorical_features:
    contingency_table = pd.crosstab(df['Kraljic_Category'], df[feature])
    chi2, p, dof, _ = chi2_contingency(contingency_table)
    n = contingency_table.values.sum()
    v_cramer = np.sqrt(chi2 / (n * (min(contingency_table.shape) - 1)))
    chi2_results.append({
        'Feature': feature,
        'Chi2': chi2,
        'P_Value': p,
        'Cramer_V': v_cramer,
        'Significant': p < 0.05
    })

chi2_df = pd.DataFrame(chi2_results)
print("\n✅ Resultados do teste Qui-Quadrado (associação com categoria):")
print(chi2_df[['Feature', 'Chi2', 'P_Value', 'Cramer_V', 'Significant']])

# ============== 3. RELATÓRIO FINAL ==============
print("\n📌 CONCLUSÕES SOBRE AS ESTRATÉGIAS DE COMPRAS ATUAIS")
print("=====================================================================================")
print("A análise revelou os seguintes padrões:\n")

# Destacar variáveis com diferença significativa
significant_anova = anova_df[anova_df['Significant']]['Feature'].tolist()
print(f"🔹 Variáveis com diferenças estatísticas significativas entre categorias: {', '.join(significant_anova)}")

high_diff_vars = category_means.std().sort_values(ascending=False).head(3).index.tolist()
print(f"\n🔹 Variáveis com maior variação entre categorias: {', '.join(high_diff_vars)}")

# Verificar associações categóricas
significant_chi2 = chi2_df[chi2_df['Significant']]['Feature'].tolist()
if significant_chi2:
    print(f"🔹 Categorias associadas estatisticamente às seguintes características: {', '.join(significant_chi2)}")
else:
    print("🔹 Nenhuma variável categórica mostrou associação estatística forte.")

# Exemplo de insights por categoria
high_lead_time_category = category_means['Lead_Time_Days'].idxmax()
avg_lead_time_by_category = category_means['Lead_Time_Days'].to_dict()
print(f"\n🔺 Produtos com maior tempo médio de entrega ({avg_lead_time_by_category[high_lead_time_category]} dias): {high_lead_time_category}")

high_cost_category = category_means['Cost_per_Unit'].idxmax()
avg_cost_by_category = category_means['Cost_per_Unit'].to_dict()
print(f"🔺 Produtos com maior custo médio (R$ {avg_cost_by_category[high_cost_category]}): {high_cost_category}")

# Recomendações finais
print("\n💡 Recomendações:")
print("- Revisar estratégias atuais com base nas variáveis com maior poder discriminatório.")
print("- Priorizar fornecedores com menor risco para produtos com alto impacto nos lucros.")
print("- Desenvolver alternativas para itens com fonte única em categorias críticas.")
print("- Padronizar processos para categorias com baixa variação entre critérios.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.font_manager import FontProperties
from matplotlib.backends.backend_pdf import PdfPages
import webbrowser
import os
import textwrap

def gerar_pdf_5w2h(data):
    # Configuração do documento PDF
    pdf_pages = PdfPages("Plano_Acão_Estratégico_5W2H.pdf")

    # Ajustes de layout - aumentando a altura para evitar sobreposição
    fig, ax = plt.subplots(figsize=(14, 14))  # Aumentei a altura para 14
    ax.axis('off')

    # Configurações de fonte
    font = FontProperties()
    font.set_size(10)
    font_bold = FontProperties()
    font_bold.set_size(11)
    font_bold.set_weight('bold')

    # Preparar os dados com quebras de linha
    wrapped_data = []
    max_lines = 1
    line_heights = []
    
    for elemento, descricao in zip(data['Elemento'], data['Descrição']):
        # Aplicar quebra de linha para descrições longas
        wrapped_desc = textwrap.fill(descricao, width=70)
        lines = wrapped_desc.count('\n') + 1
        line_heights.append(lines)
        if lines > max_lines:
            max_lines = lines
        wrapped_data.append([elemento, wrapped_desc])

    # Criação da tabela com posicionamento mais baixo
    table = ax.table(
        cellText=wrapped_data,
        colLabels=['Elemento', 'Detalhes'],
        cellLoc='left',
        loc='center',
        bbox=[0.1, 0.1, 0.8, 0.8],  # Ajuste da posição e tamanho da tabela
        colWidths=[0.2, 0.8]
    )

    # Ajustes de formatação
    table.auto_set_font_size(False)
    table.set_fontsize(10)
    
    # Ajuste dinâmico da escala baseado no número de linhas
    scale_y = 1.0 + (max_lines * 0.05)
    table.scale(1.2, scale_y)

    # Estilização das células
    for (i, j), cell in table.get_celld().items():
        cell.set_edgecolor('#4682b4')
        cell.set_linewidth(0.7)
        if i == 0:
            cell.set_facecolor('#1e3d59')
            cell.set_text_props(color='white', fontproperties=font_bold)
        else:
            # Altura dinâmica baseada no conteúdo
            cell.set_height(0.1 + (0.02 * line_heights[i-1]))
            cell.set_text_props(fontproperties=font)
            cell.PAD = 0.1

    # Título do documento com posicionamento ajustado
    plt.suptitle("Plano de Ação Estratégico (5W2H)", 
                y=0.95, fontsize=14, fontweight='bold', color='#1e3d59')

    # Finalização do layout com mais espaço
    plt.tight_layout(pad=5.0, rect=[0, 0.05, 1, 0.95])
    pdf_pages.savefig(fig, bbox_inches='tight', dpi=300)
    pdf_pages.close()
    return "Plano_Acão_Estratégico_5W2H.pdf"

def gerenciar_pdf(pdf_path):
    print("\nOpções de Visualização:")
    resposta = input("[A]brir PDF | [B]aixar PDF | [V]isualizar Tabela: ").upper()

    if resposta == 'A':
        webbrowser.open(pdf_path)
    elif resposta == 'B':
        print(f"\nArquivo disponível em:\n{os.path.abspath(pdf_path)}")
    elif resposta == 'V':
        print("\n" + "=" * 100)
        print(pd.DataFrame(data).to_string(index=False))
        print("=" * 100)
    else:
        print("Opção inválida. Por favor, tente novamente.")

# Dados atualizados do Plano de Ação Estratégico
data = {
    "Elemento": [
        "What (O quê?)",
        "Why (Por quê?)",
        "Where (Onde?)",
        "When (Quando?)",
        "Who (Quem?)",
        "How (Como?)",
        "How Much (Quanto?)"
    ],
    "Descrição": [
        "Otimizar a cadeia de suprimentos com foco em ROI, planejamento e renegociação",
        "Reduzir custos operacionais, aumentar retorno sobre compras e evitar perdas",
        "Áreas de compras, suprimentos, engenharia e manutenção",
        "Início em maio de 2025, com revisão trimestral",
        "Líder de Procurement, Analista de Dados, Equipe de Compras e CFO",
        "Aplicar classificação Kraljic, renegociar contratos, implantar planejamento trimestral e dashboard de ROI",
        "Investimento estimado de R$ 100 mil em sistemas e capacitação com retorno projetado de R$ 2,5 milhões/ano"
    ]
}

if __name__ == "__main__":
    df = pd.DataFrame(data)
    arquivo_pdf = gerar_pdf_5w2h(df)
    gerenciar_pdf(arquivo_pdf)
```
